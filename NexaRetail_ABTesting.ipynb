#### The Treatment page appears promising: it generated about a 31.6% relative increase in registrations compared with the Control page.

#### Is this observed improvement statistically significant, or could it have happened because of random variation?

```python
from statsmodels.stats.proportion import proportions_ztest
z_test=proportions_ztest(count=[137,180],nobs=[5004,4996],alternative="two-sided")
z_test
```

#### Z_score: -2.47
#### P-value: 0.01356

#### Negative z-score is not a bad thing, it just means we subtracted 180 from 137 instead of the other way, if we change that it will become positive

#### Since p_value<0.05, we reject Null hypothesis

### There is statistically significant evidence that the conversion rates of the two pages are different

## The A/B test found a statistically significant difference in conversion rates between the Control and Treatment pages. Since the Treatment page had a higher conversion rate (3.60% vs. 2.74%), we have sufficient statistical evidence to conclude that the Treatment page performed better during this experiment. 

### Observed effect: Treatment conversion is higher by about 0.86 percentage points.
### Statistical evidence: \(p = 0.0136 < 0.05\), so we reject the null hypothesis.

### The result is statistically significant and the 31.6% relative lift suggests it may also be practically meaningful. However, to confidently determine business significance, we would ideally need to know the business value of an additional registration, implementation costs, and the expected traffic volume.
